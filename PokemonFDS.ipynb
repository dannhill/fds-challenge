<h1>Create K-Fold datasets</h1>

```python
import json
import random
import os
from sklearn.model_selection import KFold

# Define file paths
input_file = './fds-challenge-dataset/train.jsonl'
output_dir = './fds-challenge-dataset/kfolds'

# Create output directory if it doesn't exist
os.makedirs(output_dir, exist_ok=True)

# Read all lines from the input file
with open(input_file, 'r') as f:
    lines = f.readlines()

# Shuffle the lines
random.shuffle(lines)

# Set up KFold
n_splits = 10
kf = KFold(n_splits=n_splits, shuffle=False) # Data is already shuffled

# Iterate through folds and create train/validation files
for fold, (train_index, val_index) in enumerate(kf.split(lines)):
    print(f"Creating fold {fold+1}/{n_splits}")

    train_lines = [lines[i] for i in train_index]
    val_lines = [lines[i] for i in val_index]

    train_output_file = os.path.join(output_dir, f'fold_{fold+1}_train.jsonl')
    val_output_file = os.path.join(output_dir, f'fold_{fold+1}_val.jsonl')

    with open(train_output_file, 'w') as f:
        for line in train_lines:
            f.write(line)

    with open(val_output_file, 'w') as f:
        for line in val_lines:
            f.write(line)

    print(f"  Fold {fold+1}: {len(train_lines)} training lines, {len(val_lines)} validation lines.")

print("Finished creating all k-fold datasets.")
```

```python
import json
import os
import torch

# --- Funzioni di caricamento e codifica (invariate) ---

# Load mappings for one-hot encoding
def load_json_keys_as_list(filepath):
    with open(filepath, 'r') as f:
        data = json.load(f)
        return list(data.keys())

try:
    # Assicurati che i percorsi siano corretti per il tuo ambiente
    base_path = './fds-challenge-dataset/'
    pokemon_names = load_json_keys_as_list(os.path.join(base_path, 'pokedex.json'))
    move_names = load_json_keys_as_list(os.path.join(base_path, 'moves.json'))
    statuses = load_json_keys_as_list(os.path.join(base_path, 'statuses.json'))
    # Aggiungiamo 'nostatus' se non è presente nel file, dato che è un valore comune
    if 'nostatus' not in statuses:
        statuses.append('nostatus')
    types = [t.lower() for t in load_json_keys_as_list(os.path.join(base_path, 'types.json'))]

    effects = ['noeffect', 'reflect'] # Assicurati che questa lista sia completa
    move_categories = ['PHYSICAL', 'SPECIAL', 'STATUS']

except FileNotFoundError as e:
    print(f"Error loading mapping file: {e}")
    raise

def one_hot_encode(value, categories, case_insensitive=False):
    """Creates a one-hot encoding vector."""
    if case_insensitive:
        value = value.lower()
        categories = [c.lower() for c in categories]

    try:
        index = categories.index(value)
        one_hot = [0.0] * len(categories)
        one_hot[index] = 1.0
        return one_hot
    except ValueError:
        # Fornisce un messaggio di errore più dettagliato
        raise ValueError(f"Value '{value}' not found in the provided categories for one-hot encoding.")

# --- Funzione di supporto per elaborare i dati di un turno per un giocatore ---

def _process_player_turn_state(turn_data, player_prefix):
    """
    Elabora lo stato e la mossa di un singolo giocatore per un turno e restituisce il suo feature vector.
    `player_prefix` dovrebbe essere 'p1' o 'p2'.
    """
    player_features = []

    pokemon_state_key = f'{player_prefix}_pokemon_state'
    move_details_key = f'{player_prefix}_move_details'

    # Calcola la dimensione del vettore per il padding in caso di dati mancanti
    # Nome(151) + HP(1) + Status(len) + Effetti(len) + Boosts(5) + Nome Mossa(165) + Tipo Mossa(20) + Cat Mossa(3) + Stat Mossa(3)
    VECTOR_SIZE_PER_PLAYER_TURN = (len(pokemon_names) + 1 + len(statuses) + len(effects) + 5 +
                                   len(move_names) + len(types) + len(move_categories) + 3)

    if turn_data.get(pokemon_state_key):
        pokemon_state = turn_data[pokemon_state_key]

        # One-hot encoding del nome del pokemon (es. 151 dimensioni)
        player_features.extend(one_hot_encode(pokemon_state['name'], pokemon_names))

        # Percentuale HP (1 dimensione numerica)
        player_features.append(float(pokemon_state['hp_pct']))

        # One-hot encoding dello status (es. 6 dimensioni)
        player_features.extend(one_hot_encode(pokemon_state['status'], statuses))

        # One-hot encoding degli effetti (es. 2 dimensioni)
        effects_one_hot = [0.0] * len(effects)
        for effect in pokemon_state['effects']:
            if effect in effects:
                effects_one_hot[effects.index(effect)] = 1.0
        player_features.extend(effects_one_hot)

        # Boosts (5 dimensioni numeriche: atk, def, spa, spd, spe)
        boost_stats = ['atk', 'def', 'spa', 'spd', 'spe']
        player_features.extend([float(pokemon_state['boosts'].get(stat, 0)) for stat in boost_stats])

        # Dettagli della mossa
        if turn_data.get(move_details_key):
            move_details = turn_data[move_details_key]

            # One-hot encoding del nome della mossa (es. 165 dimensioni)
            player_features.extend(one_hot_encode(move_details['name'], move_names))

            # One-hot encoding del tipo della mossa (es. 20 dimensioni)
            player_features.extend(one_hot_encode(move_details['type'], types, case_insensitive=True))

            # One-hot encoding della categoria della mossa (3 dimensioni)
            player_features.extend(one_hot_encode(move_details['category'], move_categories))

            # Statistiche della mossa (Potenza, Precisione, Priorità) (3 dimensioni)
            player_features.append(float(move_details.get('base_power', 0)))
            player_features.append(float(move_details.get('accuracy', 0)))
            player_features.append(float(move_details.get('priority', 0)))
        else:
            # Aggiungi zeri se non ci sono dettagli sulla mossa
            player_features.extend([0.0] * (len(move_names) + len(types) + len(move_categories) + 3))

        return player_features
    else:
        # Se non ci sono dati per il pokemon del giocatore in questo turno, aggiungi zeri
        return [0.0] * VECTOR_SIZE_PER_PLAYER_TURN


# --- Funzione principale aggiornata ---

def create_feature_vector(data_entry):
    """
    Converts a single JSON data entry into a flattened float feature vector.
    """
    feature_vector = []

    # Process P1's team details (6 Pokemon) - Questa parte rimane invariata
    for pokemon_data in data_entry['p1_team_details']:
        feature_vector.extend(one_hot_encode(pokemon_data['name'], pokemon_names))
        feature_vector.append(float(pokemon_data['level']))
        types_one_hot = [0.0] * (len(types) * 2)
        for i, pokemon_type in enumerate(pokemon_data['types']):
            if pokemon_type.lower() in types:
                type_index = types.index(pokemon_type.lower())
                if i < 2:
                    types_one_hot[type_index + i * len(types)] = 1.0
        feature_vector.extend(types_one_hot)
        feature_vector.append(float(pokemon_data['base_hp']))
        feature_vector.append(float(pokemon_data['base_atk']))
        feature_vector.append(float(pokemon_data['base_def']))
        feature_vector.append(float(pokemon_data['base_spa']))
        feature_vector.append(float(pokemon_data['base_spd']))
        feature_vector.append(float(pokemon_data['base_spe']))

    # Process P2's first Pokemon details - Questa parte rimane invariata
    if data_entry['p2_lead_details']:
        p2_pokemon1_data = data_entry['p2_lead_details']
        feature_vector.extend(one_hot_encode(p2_pokemon1_data['name'], pokemon_names))
        feature_vector.append(float(p2_pokemon1_data['level']))
        p2_types_one_hot = [0.0] * (len(types) * 2)
        for i, pokemon_type in enumerate(p2_pokemon1_data['types']):
            if pokemon_type.lower() in types:
                type_index = types.index(pokemon_type.lower())
                if i < 2:
                    p2_types_one_hot[type_index + i * len(types)] = 1.0
        feature_vector.extend(p2_types_one_hot)
        feature_vector.append(float(p2_pokemon1_data['base_hp']))
        feature_vector.append(float(p2_pokemon1_data['base_atk']))
        feature_vector.append(float(p2_pokemon1_data['base_def']))
        feature_vector.append(float(p2_pokemon1_data['base_spa']))
        feature_vector.append(float(p2_pokemon1_data['base_spd']))
        feature_vector.append(float(p2_pokemon1_data['base_spe']))
    else:
         feature_vector.extend([0.0] * (len(pokemon_names) + 1 + (len(types) * 2) + 6))

    # === SEZIONE MODIFICATA ===
    # Processa i turni dalla battle_timeline
    if 'battle_timeline' in data_entry:
        for turn_data in data_entry['battle_timeline']:
            # Processa lo stato e la mossa di P1 per questo turno
            feature_vector.extend(_process_player_turn_state(turn_data, 'p1'))

            # Processa lo stato e la mossa di P2 per questo turno
            feature_vector.extend(_process_player_turn_state(turn_data, 'p2'))

    #applica standardizzazione del feature_vector
    mean = sum(feature_vector) / len(feature_vector)
    variance = sum((x - mean) ** 2 for x in feature_vector) / len(feature_vector)
    stddev = variance ** 0.5 if variance > 0 else 1.0
    feature_vector = [(x - mean) / stddev for x in feature_vector]


    # print(f"Feature vector length: {len(feature_vector)}")  # Debug: stampa la lunghezza del vettore delle caratteristiche
    return feature_vector

# --- Esempio di utilizzo (invariato) ---

# train_file_path = '/content/drive/MyDrive/fds-challenge-dataset/kfolds/fold_1_train.jsonl'
# feature_vectors = []

# try:
#     with open(train_file_path, 'r') as f:
#         for i, line in enumerate(f):
#             try:
#                 data_entry = json.loads(line)
#                 feature_vector = create_feature_vector(data_entry)
#                 feature_vectors.append(feature_vector)
#             except (KeyError, ValueError) as e:
#                 print(f"Error processing line {i+1} in {train_file_path}: {e}")
#                 # Puoi decidere di saltare la riga o interrompere l'esecuzione
#                 continue

#     if feature_vectors:
#         vector_length = len(feature_vectors[0])
#         if all(len(v) == vector_length for v in feature_vectors):
#              feature_tensor = torch.tensor(feature_vectors, dtype=torch.float32)
#              print(f"Created a tensor with shape: {feature_tensor.shape}")
#         else:
#              print("Error: Feature vectors have inconsistent lengths.")
#              # Debug: stampa la lunghezza dei vettori anomali
#              for i, v in enumerate(feature_vectors):
#                  if len(v) != vector_length:
#                      print(f"Vector {i} has length {len(v)}, expected {vector_length}")
#     else:
#         print("No data entries were successfully processed.")

# except FileNotFoundError:
#     print(f"Error: File not found at {train_file_path}")
# except Exception as e:
#     print(f"An unexpected error occurred: {e}")
```

MLP definition and utilities

```python
import os
import json
import re
import numpy as np
import torch
from torch.utils.data import Dataset, DataLoader
import torch.nn as nn

# This constant must match the TOTAL_FEATURES macro in the C script.
# Calculation: (7 * 198) + (30 * (2 * 359)) = 1386 + 21540 = 22926
C_TOTAL_FEATURES = 22926

# In-memory cache to avoid reloading data from disk for the same file path
processed_datasets = {}

class PokemonBattleDataset(Dataset):
    """
    PyTorch Dataset for loading Pokemon battle data.
    
    This class implements a layered caching strategy:
    1. Checks an in-memory dictionary for already loaded data.
    2. Tries to load pre-processed, C-generated binary files (*_features.bin, *_labels.bin).
    3. Tries to load pre-processed, Python-generated PyTorch tensor files (*_features.pt, *_labels.pt).
    4. As a last resort, it processes the raw .jsonl file from scratch and saves it as .pt files for future runs.
    """
    def __init__(self, file_path):
        self.file_path = file_path
        
        # 1. Check in-memory cache first
        if file_path in processed_datasets:
            self.lines = processed_datasets[file_path]
            return

        # Derive base filename (e.g., "fold_1_train") from the full path
        base_name = os.path.basename(file_path).replace('.jsonl', '')
        processed_dir = './fds-challenge-dataset/kfolds/processed_datasets'
        
        # 2. Try to load from C-generated .bin files
        features_bin_path = os.path.join(processed_dir, f'{base_name}_features.bin')
        labels_bin_path = os.path.join(processed_dir, f'{base_name}_labels.bin')

        if os.path.exists(features_bin_path) and os.path.exists(labels_bin_path):
            print(f"Loading from C-generated binary files for {base_name}...")
            
            # Read the flat binary data using numpy
            features_flat = np.fromfile(features_bin_path, dtype=np.float32)
            labels_flat = np.fromfile(labels_bin_path, dtype=np.float32)
            
            num_samples = len(labels_flat)
            
            # Reshape the features array into [num_samples, feature_size]
            features_reshaped = features_flat.reshape(num_samples, C_TOTAL_FEATURES)
            
            # Convert numpy arrays to PyTorch tensors
            features_tensor = torch.from_numpy(features_reshaped)
            labels_tensor = torch.from_numpy(labels_flat).view(-1, 1) # Reshape labels to [N, 1]
            
            self.lines = list(zip(features_tensor, labels_tensor))
            processed_datasets[file_path] = self.lines
            print(f"Successfully loaded {len(self.lines)} samples from .bin files.")
            return

        # 3. Fallback: Try to load from Python-generated .pt files
        features_pt_path = os.path.join(processed_dir, f'{base_name}_features.pt')
        labels_pt_path = os.path.join(processed_dir, f'{base_name}_labels.pt')

        if os.path.exists(features_pt_path) and os.path.exists(labels_pt_path):
            print(f"Loading from Python-generated .pt files for {base_name}...")
            with open(features_pt_path, 'rb') as f_features, open(labels_pt_path, 'rb') as f_labels:
                features = torch.load(f_features)
                labels = torch.load(f_labels)
                self.lines = list(zip(features, labels))
                processed_datasets[file_path] = self.lines
                print(f"Successfully loaded {len(self.lines)} samples from .pt files.")
                return
        
        # 4. Final Fallback: Process from the source .jsonl file
        # NOTE: This requires the 'create_feature_vector' function to be defined elsewhere in your script.
        print(f"No pre-processed files found. Processing from source: {file_path}")
        os.makedirs(processed_dir, exist_ok=True)
        
        with open(file_path, 'r') as f:
            original_lines = f.readlines()
            
        processed_data = []
        for idx, line in enumerate(original_lines):
            # This print statement can be verbose, consider removing for large files
            # print(f"Processing line {idx + 1}/{len(original_lines)}...")
            parsed_line = json.loads(line)
            # The 'create_feature_vector' function is assumed to exist
            feature_vec = create_feature_vector(parsed_line) 
            label = float(bool(parsed_line["player_won"]))
            processed_data.append(
                (torch.tensor(feature_vec, dtype=torch.float32),
                 torch.tensor([label], dtype=torch.float32))
            )
            
        self.lines = processed_data
        processed_datasets[file_path] = self.lines
        
        # Save features and labels as .pt files for faster future loading
        features = torch.stack([feat for feat, label in self.lines])
        labels = torch.stack([label for feat, label in self.lines])
        
        with open(features_pt_path, 'wb') as f_features:
            torch.save(features, f_features)
        with open(labels_pt_path, 'wb') as f_labels:
            torch.save(labels, f_labels)
        print(f"Finished processing and saved to .pt files for future use.")


    def __len__(self):
        return len(self.lines)

    def __getitem__(self, idx):
        features_tensor, label_tensor = self.lines[idx]
        return features_tensor, label_tensor

# The PreprocessedPokemonDataset class is now somewhat redundant if you only use .bin files,
# but it's a good, fast loader for .pt files specifically.
class PreprocessedPokemonDataset(Dataset):
    """
    A very fast PyTorch Dataset for loading pre-processed tensor data saved with torch.save().
    """
    def __init__(self, features_path, labels_path):
        # Load the entire dataset into RAM. This is extremely fast.
        self.features = torch.load(features_path)
        self.labels = torch.load(labels_path)
        print(f"Loaded pre-processed data from {os.path.basename(features_path)}. Features shape: {self.features.shape}")

    def __len__(self):
        return self.features.shape[0]

    def __getitem__(self, idx):
        # Indexing a tensor in memory is orders of magnitude faster than parsing JSON.
        return self.features[idx], self.labels[idx]


class BattleMLP(nn.Module):
    def __init__(self, input_size, hidden_size1=512, hidden_size2=256):
        super(BattleMLP, self).__init__()

        self.network = nn.Sequential(
            nn.Linear(input_size, hidden_size1),
            nn.ReLU(),
            nn.Dropout(0.5),
            nn.Linear(hidden_size1, hidden_size2),
            nn.ReLU(),
            nn.Dropout(0.5),
            nn.Linear(hidden_size2, 1),
            nn.Sigmoid()
        )

    def forward(self, x):
        return self.network(x)

# Example of creating a dataset (we will do this in the training cell)
# Assume 'create_feature_vector' is defined and the necessary JSON files are in place.
# Also assume the C script has been run to generate the .bin files.

# dummy_file = './fds-challenge-dataset/kfolds/fold_1_train.jsonl'
# if os.path.exists(dummy_file):
#     train_dataset = PokemonBattleDataset(dummy_file)
#     print(f"Successfully created dataset with {len(train_dataset)} samples.")
#     if len(train_dataset) > 0:
#         features, label = train_dataset[0]
#         print(f"Sample feature shape: {features.shape}")
#         print(f"Sample label shape: {label.shape}")

#         # Example of creating a model instance
#         model = BattleMLP(input_size=features.shape[0])
#         print(model)
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
from tqdm import tqdm
import ipywidgets as widgets
from IPython.display import display
import os
import optuna

# Define the directory where checkpoints are saved
checkpoint_dir = './pokemon_battle_checkpoints'
os.makedirs(checkpoint_dir, exist_ok=True) # Ensure directory exists

# Get a list of available checkpoint files
try:
    checkpoint_files = [f for f in os.listdir(checkpoint_dir) if f.endswith('.pth')]
    checkpoint_files.sort() # Sort for consistent order
except FileNotFoundError:
    print(f"Warning: Checkpoint directory not found at {checkpoint_dir}")
    checkpoint_files = []

if not checkpoint_files:
    print("No checkpoints found. Training will start from scratch.")

# Create a dropdown widget for selecting a checkpoint
checkpoint_dropdown = widgets.Dropdown(
    options=['None'] + checkpoint_files, # Add 'None' option to start training from scratch
    description='Select Checkpoint:',
    disabled=False,
)

# --- Hyperparameters and Setup ---
# LEARNING_RATE will be determined by Optuna
BATCH_SIZE = 256
EPOCHS = 4  # Train for 5 epochs as requested
INPUT_DIMENSION = 22926
NUM_FOLDS = 1 # Total number of folds
N_TRIALS_OPTUNA = 30 # Number of trials for Optuna search

# Setup device
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")

# --- Optuna Objective Function ---
def objective(trial):
    """
    Objective function for Optuna to optimize.
    Trains the model for ONE epoch using k-fold validation and returns the
    average validation accuracy.
    """
    # Suggest hyperparameters
    lr = trial.suggest_float("lr", 1e-5, 1e-1, log=True)

    # Initialize Model, Loss, and Optimizer for this trial
    model = BattleMLP(input_size=INPUT_DIMENSION).to(device)
    loss_fn = nn.BCELoss()
    optimizer = optim.Adam(model.parameters(), lr=lr)

    epoch_val_accs = []

    # --- K-Fold Cross-Training Loop for one epoch ---
    for fold in range(1, NUM_FOLDS + 1):
        print(f"\n--- Trial {trial.number}, Fold {fold}/{NUM_FOLDS} ---")

        # Prepare DataLoaders
        train_file = f'./fds-challenge-dataset/kfolds/fold_{fold}_train.jsonl'
        val_file = f'./fds-challenge-dataset/kfolds/fold_{fold}_val.jsonl'

        try:
            train_dataset = PokemonBattleDataset(train_file)
            val_dataset = PokemonBattleDataset(val_file)
        except FileNotFoundError as e:
            print(f"Error loading data for fold {fold}: {e}. Skipping this fold.")
            continue

        train_loader = DataLoader(dataset=train_dataset, batch_size=BATCH_SIZE, shuffle=True)
        val_loader = DataLoader(dataset=val_dataset, batch_size=BATCH_SIZE, shuffle=False)

        # --- Training Phase ---
        model.train()
        for features, labels in tqdm(train_loader, desc=f"Trial {trial.number} [Fold {fold} Train]", leave=False):
            features, labels = features.to(device), labels.to(device)
            outputs = model(features)
            loss = loss_fn(outputs, labels)
            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

        # --- Validation Phase ---
        model.eval()
        val_correct, val_total = 0, 0
        with torch.no_grad():
            for features, labels in val_loader:
                features, labels = features.to(device), labels.to(device)
                outputs = model(features)
                predicted = (outputs > 0.5).float()
                val_total += labels.size(0)
                val_correct += (predicted == labels).sum().item()

        fold_val_acc = 100 * val_correct / val_total
        epoch_val_accs.append(fold_val_acc)

    # Return the average validation accuracy across all folds
    avg_val_acc = sum(epoch_val_accs) / len(epoch_val_accs) if epoch_val_accs else 0
    return avg_val_acc


# --- Function to start training ---
def start_training(checkpoint_filename):
    # --- 1. Hyperparameter Optimization with Optuna ---
    print("===== Starting Hyperparameter Optimization with Optuna =====")
    # sampler = optuna.samplers.RandomSampler()
    # study = optuna.create_study(direction='maximize', sampler=sampler)
    # study.optimize(objective, n_trials=N_TRIALS_OPTUNA)

    # best_lr = study.best_trial.params['lr']
    # print(f"\nOptuna study finished.")
    # print(f"  Best trial: {study.best_trial.number}")
    # print(f"  Best validation accuracy (avg over folds): {study.best_value:.2f}%")
    # print(f"  Best learning rate: {best_lr:.6f}")
    # print("========================================================\n")

    best_lr = 1e-4 # Hardcoded best LR from previous runs for demonstration

    # --- 2. Full Training with Best Hyperparameters ---
    print(f"===== Starting Full Training for {EPOCHS} Epochs with LR={best_lr:.6f} =====")
    # Initialize Model, Loss, and Optimizer with the best LR
    model = BattleMLP(input_size=INPUT_DIMENSION).to(device)
    loss_fn = nn.BCELoss()
    optimizer = optim.Adam(model.parameters(), lr=best_lr)

    # Load checkpoint if selected
    if checkpoint_filename != 'None':
        checkpoint_path = os.path.join(checkpoint_dir, checkpoint_filename)
        try:
            model.load_state_dict(torch.load(checkpoint_path, map_location=device))
            print(f"Loaded model from checkpoint: {checkpoint_path}")
        except Exception as e:
            print(f"Error loading checkpoint {checkpoint_path}: {e}. Starting training from scratch.")

    # --- Main Training Loop ---
    for epoch in range(EPOCHS):
        print(f"\n========================= EPOCH {epoch+1}/{EPOCHS} =========================")
        epoch_train_losses, epoch_train_accs = [], []
        epoch_val_losses, epoch_val_accs = [], []

        # --- K-Fold Cross-Training Loop ---
        for fold in range(1, NUM_FOLDS + 1):
            print(f"\n--- Processing Epoch {epoch+1}, Fold {fold}/{NUM_FOLDS} ---")
            train_file = f'./fds-challenge-dataset/kfolds/fold_{fold}_train.jsonl'
            val_file = f'./fds-challenge-dataset/kfolds/fold_{fold}_val.jsonl'

            try:
                train_dataset = PokemonBattleDataset(train_file)
                val_dataset = PokemonBattleDataset(val_file)
            except FileNotFoundError as e:
                print(f"Error loading data for fold {fold}: {e}. Skipping this fold.")
                continue

            train_loader = DataLoader(dataset=train_dataset, batch_size=BATCH_SIZE, shuffle=True)
            val_loader = DataLoader(dataset=val_dataset, batch_size=BATCH_SIZE, shuffle=False)

            # --- Training Phase ---
            model.train()
            train_loss, train_correct, train_total = 0.0, 0, 0
            train_pbar = tqdm(train_loader, desc=f"Epoch {epoch+1} [Fold {fold} Train]")
            for features, labels in train_pbar:
                features, labels = features.to(device), labels.to(device)
                outputs = model(features)
                loss = loss_fn(outputs, labels)
                optimizer.zero_grad()
                loss.backward()
                optimizer.step()
                train_loss += loss.item()
                predicted = (outputs > 0.5).float()
                train_total += labels.size(0)
                train_correct += (predicted == labels).sum().item()
                train_pbar.set_postfix({'loss': f'{train_loss/(train_pbar.n+1):.4f}', 'acc': f'{100*train_correct/train_total:.2f}%'})

            epoch_train_losses.append(train_loss / len(train_loader))
            epoch_train_accs.append(100 * train_correct / train_total)

            # --- Validation Phase ---
            model.eval()
            val_loss, val_correct, val_total = 0.0, 0, 0
            val_pbar = tqdm(val_loader, desc=f"Epoch {epoch+1} [Fold {fold} Val]")
            with torch.no_grad():
                for features, labels in val_pbar:
                    features, labels = features.to(device), labels.to(device)
                    outputs = model(features)
                    loss = loss_fn(outputs, labels)
                    val_loss += loss.item()
                    predicted = (outputs > 0.5).float()
                    val_total += labels.size(0)
                    val_correct += (predicted == labels).sum().item()
                    val_pbar.set_postfix({'loss': f'{val_loss/(val_pbar.n+1):.4f}', 'acc': f'{100*val_correct/val_total:.2f}%'})

            epoch_val_losses.append(val_loss / len(val_loader))
            epoch_val_accs.append(100 * val_correct / val_total)

        # --- Epoch Summary ---
        if not epoch_val_accs:
            print("No data was processed in this epoch. Cannot generate summary.")
            continue

        avg_epoch_train_loss = sum(epoch_train_losses) / len(epoch_train_losses)
        avg_epoch_train_acc = sum(epoch_train_accs) / len(epoch_train_accs)
        avg_epoch_val_loss = sum(epoch_val_losses) / len(epoch_val_losses)
        avg_epoch_val_acc = sum(epoch_val_accs) / len(epoch_val_accs)

        print(f"\n---------- SUMMARY FOR EPOCH {epoch+1} ----------")
        print(f"  Average Training Loss (all folds):   {avg_epoch_train_loss:.4f}")
        print(f"  Average Training Accuracy (all folds): {avg_epoch_train_acc:.2f}%")
        print(f"  Average Validation Loss (all folds):   {avg_epoch_val_loss:.4f}")
        print(f"  Average Validation Accuracy (all folds): {avg_epoch_val_acc:.2f}%")
        print("-------------------------------------------\n")

    print("Training finished.")
    checkpoint_path = os.path.join(checkpoint_dir, f'battle_mlp_final_model.pth')
    torch.save(model.state_dict(), checkpoint_path)
    print(f"Model checkpoint saved to: {checkpoint_path}")

# Display the dropdown and a button to start training
start_button = widgets.Button(description="Find Best LR & Start Training")
output = widgets.Output()

def on_button_clicked(b):
    with output:
        output.clear_output() # Clear previous output
        selected_checkpoint = checkpoint_dropdown.value
        start_training(selected_checkpoint)

start_button.on_click(on_button_clicked)

display(checkpoint_dropdown, start_button, output)
```

```python
import os

# Define the directory where the processed files are located
processed_dir = './fds-challenge-dataset/kfolds/processed_datasets'

# Define the fold number to merge
fold_to_merge = 4

# Define the input and output file names
files_to_merge = {
    'features': {
        'train': f'fold_{fold_to_merge}_train_features.bin',
        'val': f'fold_{fold_to_merge}_val_features.bin',
        'out': 'train_features.bin'
    },
    'labels': {
        'train': f'fold_{fold_to_merge}_train_labels.bin',
        'val': f'fold_{fold_to_merge}_val_labels.bin',
        'out': 'train_labels.bin'
    }
}

# Function to merge two binary files into one
def merge_files(file_paths, output_path):
    """Merges a list of binary files into a single output file."""
    try:
        with open(output_path, 'wb') as outfile:
            for f_path in file_paths:
                if os.path.exists(f_path):
                    with open(f_path, 'rb') as infile:
                        outfile.write(infile.read())
                    print(f"  - Appended {os.path.basename(f_path)}")
                else:
                    print(f"Warning: Input file not found, skipping: {f_path}")
        print(f"Successfully created merged file: {output_path}\n")
    except IOError as e:
        print(f"An error occurred while merging files into {output_path}: {e}")

# Process both features and labels
for file_type, paths in files_to_merge.items():
    train_path = os.path.join(processed_dir, paths['train'])
    val_path = os.path.join(processed_dir, paths['val'])
    output_path = os.path.join(processed_dir, paths['out'])
    
    print(f"Merging {file_type} for fold {fold_to_merge}...")
    
    # List of files to merge in order
    input_files = [train_path, val_path]
    
    # Call the merge function
    merge_files(input_files, output_path)

print("All specified files have been merged.")
```
